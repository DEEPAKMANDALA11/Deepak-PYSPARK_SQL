```python
!pip install nltk
from pyspark.sql import SparkSession
from pyspark.sql.functions import rand
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from nltk.corpus import stopwords
import nltk

nltk.download('stopwords')

# Create SparkSession
spark = SparkSession.builder.appName("CustomerSampling").getOrCreate()

# Sample social media dataset
social_data = [
    (1, "Just bought a new phone!", "2025-03-20"),
    (2, "Love this new coffee shop!", "2025-03-19"),
    (3, "Excited about the new product launch!", "2025-03-18"),
    (4, "Having a great time at the concert!", "2025-03-17"),
    (5, "Best movie of the year!", "2025-03-16"),
]

social_df = spark.createDataFrame(social_data, ["customer_id", "post", "date"])

# Sample transactional history dataset
transaction_data = [
    (1, "Electronics", 1200.50, "2025-03-15"),
    (2, "Coffee", 5.75, "2025-03-14"),
    (3, "Groceries", 250.30, "2025-03-13"),
    (4, "Concert Ticket", 150.00, "2025-03-12"),
    (5, "Movie Ticket", 12.00, "2025-03-11"),
]

transaction_df = spark.createDataFrame(transaction_data, ["customer_id", "category", "amount", "date"])

# Join datasets
compdf = transaction_df.join(social_df, on="customer_id", how="inner")

# Function to generate recommendations for a specific customer ID
def generate_recommendations(customer_id, top_n=3):  # Add top_n parameter
    # Filter data for the specific customer ID
    customer_data = compdf.filter(compdf.customer_id == customer_id).collect()

    if not customer_data:
        print(f"No data found for customer ID: {customer_id}")
        return

    # Extract posts, categories, and amounts for the customer
    post = [row.post for row in customer_data]
    TCAT = [row.category for row in customer_data]
    TAmnt = [row.amount for row in customer_data]

    # Preprocess social media posts
    stop_words = set(stopwords.words('english'))
    vectorizer = TfidfVectorizer(stop_words=list(stop_words))
    social_media_features = vectorizer.fit_transform(post)

    # Sample product data
    products = [
        {"name": "Travel Insurance", "category": "travel"},
        {"name": "Credit Card with Travel Rewards", "category": "banking"},
        {"name": "Laptop", "category": "electronics"},
        {"name": "Fine Dining Voucher", "category": "dining"},
        {"name": "Movies List", "category": "Entertainment"},
    ]

    # Combine all categories
    all_categories = list(set(TCAT + [p['category'] for p in products]))
    all_features = list(vectorizer.get_feature_names_out()) + all_categories

    # Convert features to DataFrame
    features = pd.DataFrame(social_media_features.toarray(), columns=vectorizer.get_feature_names_out())
    features = pd.concat([features, pd.get_dummies(TCAT)], axis=1)
    features = features.reindex(columns=all_features, fill_value=0)

    # Prepare product features
    product_features = []
    for product in products:
        product_categories = [1 if product['category'] == cat else 0 for cat in all_categories]
        product_features.append(product_categories)

    product_features = pd.DataFrame(product_features, columns=all_categories)
    product_features = product_features.reindex(columns=all_features, fill_value=0)

    # Calculate similarity between customer and products
    similarity_scores = cosine_similarity(features, product_features)

    # Generate top N recommendations
    customer_recommendations = []
    for i, p in enumerate(post):
        # Get indices of top N similarity scores
        top_product_indices = similarity_scores[i].argsort()[-top_n:][::-1]
        recommendations = [products[idx]['name'] for idx in top_product_indices]
        customer_recommendations.append({
            "post": p,
            "recommended_products": recommendations  # Store multiple recommendations
        })

    # Display recommendations
    print(f"Recommendations for Customer ID: {customer_id}")
    for rec in customer_recommendations:
        print(f"Post: {rec['post']}")
        print(f"Top {top_n} Recommended Products: {', '.join(rec['recommended_products'])}")
        print("-" * 40)

# Example usage: Pass customer ID and number of recommendations as input
customer_id = int(input("Enter Customer ID: "))
top_n = int(input("Enter number of recommendations (e.g., 3): "))
generate_recommendations(customer_id, top_n)
```
